```python
# Cell 2 — Upload Tomography.zip and extract
from google.colab import files
import zipfile, sys, os

uploaded = files.upload()          # pick Tomography.zip from your Desktop
zip_name = list(uploaded.keys())[0]

with zipfile.ZipFile(zip_name) as z:
    z.extractall('/content/')

# Auto-detect extracted folder
for folder in ['/content/Tomography', '/content/shadowvqe']:
    if os.path.isdir(folder + '/src'):
        sys.path.insert(0, folder + '/src')
        ROOT = folder
        break

os.makedirs(ROOT + '/figures', exist_ok=True)
print(f"Extracted to: {ROOT}")
```
